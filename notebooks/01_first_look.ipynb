## How the M5 tables connect

### sales_train_validation
Each row represents an item sold in a specific store.
The columns d_1, d_2, ..., represent daily unit sales.

### calendar
Maps each d_x value to an actual calendar date and contains
additional information such as week, events and SNAP indicators.

sales.d_x → calendar.d

### sell_prices
Contains the weekly selling price for each item in each store.

The connection is based on:
- store_id
- item_id
- wm_yr_wk

calendar provides the wm_yr_wk needed to connect daily sales
to the corresponding weekly price.

```python
# Melt sales from wide to long
id_vars = ['id', 'item_id', 'dept_id', 'cat_id', 'store_id', 'state_id']
sales_long = sales.melt(
    id_vars=id_vars,
    var_name='d',
    value_name='sold'
)

# Merge with calendar to get real dates
sales_long = sales_long.merge(
    calendar[['d', 'date', 'wm_yr_wk', 'weekday', 'month', 'year',
              'event_name_1', 'event_type_1', 'snap_CA', 'snap_TX', 'snap_WI']],
    on='d', how='left'
)
sales_long['date'] = pd.to_datetime(sales_long['date'])

print(sales_long.shape)  
sales_long.head()
```
